```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()

local_api_key = os.environ["LOCAL_LLM_API_KEY"]
local_base_url = os.environ["LOCAL_LLM_BASE_URL"]

def get_local_llm(model, **kwargs):
    return ChatGoogleGenerativeAI(
        model=model,
        api_key=local_api_key,
        base_url=local_base_url,
        **kwargs,
    )
```

## Call LLM Using Google LLM API

```python
llm = get_local_llm(model="gemini-auto")

response = llm.invoke("How many moons does Jupiter have?")
print(response.text)
```

```python
llm = get_local_llm(model="gemini-auto")

response = llm.invoke([
    ["system", "You are a helpful assistant that answers in short one line"],
    ["human", "How many moons does Jupiter have?"]
])
print(response.text)
```

### Temperature 


```python
prompt = "Write a single tagline for a quirky coffee shop. Only output the tagline, nothing else."

llm = get_local_llm(model="gemini-auto", temperature=0)

response = llm.invoke(prompt)
print(response.text)
```

```python
llm = get_local_llm(model="gemini-auto", temperature=1)

response = llm.invoke(prompt)
print(response.text)
```

## Call LLM Using Groq Models

```python
from langchain_groq import ChatGroq
```

```python
llm = ChatGroq(model="qwen/qwen3-32b", reasoning_format="parsed")

messages = [
    (
        "system",
        "You are a helpful assistant that gives a short answer to a question",
    ),
    (
        "human", 
        "How many moons does Jupiter have?"
    ),
]
ai_msg = llm.invoke(messages)
print(ai_msg.content)
```
